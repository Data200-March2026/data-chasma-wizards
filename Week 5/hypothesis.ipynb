**Hypothesis Testing**


H₀: Independent variables have no effect on lung cancer


H₁: Independent variables significantly affects lung cancer

### Step 1: Importing Required Libraries

- **pandas**: Helps us work with data in tables (like Excel).
- **numpy**: Useful for working with numbers, and calculations.
- **matplotlib.pyplot**: Lets us make charts, and graphs.
- **seaborn**: Makes our charts look nicer, and easier to read.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2: Loading the Data

- **pd.read_csv()**: Reads the CSV file (like an Excel sheet), and puts it into a table called a DataFrame.
- **df.head()**: Shows us the first few rows of the data, so we can quickly check what it looks like.

This step is important to make sure our data is loaded correctly before we start any analysis.

```python
df = pd.read_csv("../Dataset/cleaned_lcs.csv")
df.head()
```

#### Output Explanation (Step 2: Data Preview)

- This table shows the first few rows of the dataset after loading it.
- Each row is a person, and each column is a feature (like age, smoking, allergy, etc.).
- Values are 0 or 1 for yes/no questions, and numbers for age.
- This helps us quickly check that the data loaded correctly, and see what kind of information we have.

### Step 3: Logistic Regression Model
Here, we build a model to see which factors affect lung cancer:

- **import statsmodels.api as sm**: Loads a library for advanced statistics, and modeling.
- **X = df.drop(columns=['lung_cancer'])**: We take all columns except 'lung_cancer' as our input features (things that might affect the result).
- **y = df['lung_cancer']**: This is our target variable (what we want to predict).
- **sm.add_constant(X)**: Adds a column of 1s to our data, which helps the model work better.
- **sm.Logit(y, X).fit()**: Fits (trains) a logistic regression model to our data. This helps us see which features are important.
- **model.summary()**: Shows a detailed report of the model results, including which features matter most.


```python
import statsmodels.api as sm
#preparing data for logistic regression
X = df.drop(columns=['lung_cancer'])
y = df['lung_cancer']
#adding constant
X = sm.add_constant(X)
#fitting model
model = sm.Logit(y, X).fit()
# print model summary
print(model.summary())
```

#### Output Explanation (Step 3: Logistic Regression Results)

- The table shows which factors (like smoking, anxiety, allergy, etc.) affect lung cancer the most.
- **coef**: Positive values mean higher risk, negative mean lower risk.
- **P>|z|**: If less than 0.05, that factor is important (statistically significant).
- For example, smoking, anxiety, allergy have low p-values, so they are important.
- The warning at the bottom means some answers in the data are very clear, so the model finds it easy to predict those cases.

**Conclusion: We REJECT the null hypothesis**
- Since several variables (like smoking, anxiety, allergy) have p-values < 0.05, they are statistically significant.
- This means the independent variables DO significantly affect lung cancer.
- The alternative hypothesis (H₁) is supported.

### Step 4: Checking Categorical Feature Balance
Looking at how the different categories (like yes/no answers) are spread out in our data:

- **exclude_cols**: We make a list of columns we don't want to check right now (like age, gender, etc.).
- **categorical_cols**: We find all the other columns that are categories (like yes/no questions).
- **math.ceil()**: Helps us decide how many rows of charts we need.
- **plt.figure()**: Sets up the space for our charts.
- **sns.countplot()**: Makes a bar chart for each category, showing how many people answered yes or no.
- **plt.tight_layout() & plt.show()**: Makes sure the charts look neat, and are easy to see.

This step helps us see if our data is balanced, or if some answers are much more common than others.

```python
import math
# removing columns with p-values more than 0.05
exclude_cols = ['age', 'gender', 'peer_pressure', 'yellow_fingers', 'chronic disease', 'fatigue']
categorical_cols = [col for col in df.columns if col not in exclude_cols]
n_cols = 5
n_rows = math.ceil(len(categorical_cols) / n_cols)
plt.figure(figsize=(n_cols * 5, n_rows * 4))
plt.suptitle('Binary Feature Distribution', fontsize=16, y=1.02)
for idx, col in enumerate(categorical_cols, 1):
    plt.subplot(n_rows, n_cols, idx)
    ax = sns.countplot(x=col, data=df)
    plt.title(f'{col}')
    plt.xlabel('')
    plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

#### Output Explanation (Step 4: Binary Feature Distribution)

- Each chart shows how many people answered "yes" (1), or "no" (0) for each feature (like smoking, anxiety, allergy, etc.).
- If the bars are similar in height, the data is balanced for that feature.
- This helps us see if we have enough examples of each answer for every question.

### Step 5: Checking Target Class Balance
Now, checking how many people in our data have lung cancer, and how many do not:

- **df['lung_cancer'].value_counts()**: Counts how many people have each value (yes or no) for lung cancer.
- **sns.countplot()**: Makes a bar chart to show the balance between the two groups.
- **plt.title() & plt.show()**: Adds a title, and displays the chart.

This step is important to check if our data is fair, or if one group is much bigger than the other.

```python
# Class Balance Check
print("\n--- Lung Cancer Class Balance ---")
print(df['lung_cancer'].value_counts())
plt.figure(figsize=(6, 4))
sns.countplot(x='lung_cancer', data=df)
plt.title('Lung Cancer Class Balance')
plt.show()
```

#### Output Explanation (Step 5: Lung Cancer Class Balance)

- The chart shows how many people in the data have lung cancer (1), and how many do not (0).
- The bars are almost equal, which means the data is balanced.
- This is good because it helps the model learn from both groups fairly.
